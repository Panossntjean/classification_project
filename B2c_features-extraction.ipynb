```python
import cv2
import numpy as np
import os
```

```python
# Εξαγωγή χαρακτηριστικών από την κάθε εικόνα
def extract_features(img_path):
    img = cv2.imread(img_path)
    img = cv2.resize(img, (200, 200))

    R, G, B = img[:,:,2], img[:,:,1], img[:,:,0]

    # RGB στατιστικά
    r_mean, g_mean, b_mean = R.mean(), G.mean(), B.mean()
    r_std, g_std, b_std = R.std(), G.std(), B.std()
    
    # Brightness
    brightness = np.mean(img)

    # Cloudiness index
    cloud_index = b_mean / (r_mean + g_mean + 1)

    # Edge density
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    edges = cv2.Canny(gray, 50, 150)
    edge_density = np.sum(edges > 0) / (200 * 200)

    return [r_mean, g_mean, b_mean, r_std, g_std, b_std, brightness, cloud_index, edge_density]
```

```python
def load_dataset(root_dir):
    features, labels = [], []
    for label, folder in enumerate(["sunny", "cloudy"]):
        path = os.path.join(root_dir, folder)
        for img_file in os.listdir(path):
            if img_file.endswith(".jpg") or img_file.endswith(".png"):
                full_path = os.path.join(path, img_file)
                feats = extract_features(full_path)
                features.append(feats)
                labels.append(label)
    return np.array(features), np.array(labels)
```

```python
X, y = load_dataset("data/train/")
```

Αφαίρεση χαρακτηριστικών με χαμηλή διακύμανση

```python
from sklearn.feature_selection import VarianceThreshold
```

```python
selector = VarianceThreshold(threshold=0.01)
X_selected = selector.fit_transform(X)
```
